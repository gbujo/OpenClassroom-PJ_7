# Tester l'API Predict en local

```python
import requests

#url = "http://localhost:8080/api/predict"  # Serveur Flask local
url = 'https://predictocpj7-1013078366791.europe-west9.run.app/api/predict'  # On GCP 

domain_features=['DAYS_BIRTH', 'CODE_GENDER_F', 'FLAG_RISKED_ORGANIZATION_TYPE', 
                 'RATIO_INCOME_TO_FAM_MEMBERS', 'FLAG_STABLE_INCOME_TYPE',
                 'EXT_SOURCE_2',
                 'AMT_ALL_SUM_MEAN', 'RATIO_ALL_SUM_DEBT_TO_INCOME', 'RATIO_ALL_SUM_OVERDUE_TO_INCOME', 'RATIO_ALL_MAX_OVERDUE_TO_INCOME']

client_data = {
    'DAYS_BIRTH': -6.0, 
    'CODE_GENDER_F': 1, 
    'FLAG_RISKED_ORGANIZATION_TYPE': 1, 
    'RATIO_INCOME_TO_FAM_MEMBERS': 9999.0, 
    'FLAG_STABLE_INCOME_TYPE': 1,
    'EXT_SOURCE_2': 0.4,
    'AMT_ALL_SUM_MEAN': 10000.0, 
    'RATIO_ALL_SUM_DEBT_TO_INCOME': 0.20, 
    'RATIO_ALL_SUM_OVERDUE_TO_INCOME': 0.0, 
    'RATIO_ALL_MAX_OVERDUE_TO_INCOME': 0.0
}

response = requests.post(url, json=client_data)

if response.status_code == 200:
    print("Requête POST réussie")
    print(response.text)
    print("Contenu de la réponse :", response.json())
else:
    print("Erreur lors de la requête POST")
    print("Code de statut :", response.status_code)
    print("Contenu de la réponse :", response.text) 
```
